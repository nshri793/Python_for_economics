```python
#.loc — label-based selection, selects rows and columns based on their labels/names.
#general synatx: df.loc[row_label, column_label]
df.loc[1, 'Name'] #Go to the row whose label is 1 and the column whose label is Name."
df.loc[1] # selecting a complete row
df.loc[[0, 2]] #selects rows with labels 0 and 2
df.loc[:, 'Name'] #selecting columns using .loc; Select all rows and the column called Name.
df.loc[:, ['Name', 'Income']] #multiple columns
df.loc[1:3] #.loc with a range-including the starting and ending labels
```

```python
# .iloc — position-based selection, integer location, physical position of the row or column
df.iloc[row_position, column_position] #general syntax
df.iloc[1,0]
df.iloc[1] #selecting rows with .iloc
df.iloc[[0, 2]] #selecting multiple rows with .iloc
df.iloc[:, 0] #selecting columns with .iloc
df.iloc[:, 1:3] #selecting columns with a range .iloc
```

```python
#difference between .loc and .iloc
df.loc[1:3] #the ending value is included, inclusive
df.iloc[1:3] #the ending value is excluded, exclusive, useful when you care baout positions
```

```python
#Combining datasets: Merge, join and Concatenate. They all require a common variable prefreable an ID/key
#merge() is used when two datasets need to be combined based on a common key/column.Match these rows based on ID.
#Join() is primarily used to combine DataFrames based on their index, although it can also join using columns.Match these rows using their index.
#concat() is used to stack datasets. Put these datasets together one after another or side by side.
```

```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [101, 102, 103, 104],
    'Name': ['Amit', 'Neha', 'Rahul', 'Priya'],
    'Gender': ['Male', 'Female', 'Male', 'Female']
})

df2 = pd.DataFrame({
    'ID': [101, 102, 103, 104],
    'Income': [25000, 35000, 28000, 40000],
    'Education': ['Graduate', 'Postgraduate',
                  'Graduate', 'Postgraduate']
})
```

```python
df1
```

```python
df2
```

```python
df = pd.merge(df1, df2, on='ID')
```

```python
df
```

```python
#Types of merge:inner, left, right and outer joins
```

```python
#Inner merge: Keep only observations present in both datasets.inner=intersection
pd.merge(
    df1,
    df2,
    on='ID',
    how='inner'
)
```

```python
#left merge: Keep every observation from the left DataFrame (df1).
pd.merge(
    df1,
    df2,
    on='ID',
    how='left'
)
```

```python
#right merge: preserve everything in the right dataset
pd.merge(
    df1,
    df2,
    on='ID',
    how='right'
)
```

```python
#Outer merge: This keeps everything from both datasets. Outer=union, all ids from both datasets are retained
pd.merge(
    df1,
    df2,
    on='ID',
    how='outer'
)
```

```python
#bydefault merge-inner
pd.merge(df1, df2, on='ID')
```

```python
#if in case the common variable has different names
pd.merge(
    df1,
    df2,
    left_on='Person_ID',
    right_on='ID'
)
```

```python
#if in case there are multiple keys/identifiers such as state, district
pd.merge(
    df1,
    df2,
    on=['State', 'District']
)
```

```python
#Validating : type of relationship a. one to one, b.one to many, c.many to one, d. many to many
pd.merge(
    df1,
    df2,
    on='ID',
    validate='one_to_one'
)
```

```python
#how to spot duplicate keys
df1['ID'].duplicated().sum()
df2['ID'].duplicated().sum()
```

```python
#Find the duplicate ids
df2.loc[
    df2['ID'].duplicated(keep=False)
].sort_values('ID')
```

```python
#checking for a merge, this generates _merge column in the dataset with values as left only, light only, both
merged = pd.merge(
    df1,
    df2,
    on='ID',
    how='outer',
    indicator=True
)
```

```python
#finding unmatched observations
merged[merged['_merge'] == 'left_only'] #observations present only in DF1
merged[merged['_merge'] == 'right_only'] #observations present only in DF2
merged[merged['_merge'] == 'both'] #observations present in DF1 & DF2
```
